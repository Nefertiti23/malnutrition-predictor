## Analysing Weight for Age Z-Score
The waz_score variable measures how far a child's weight is from the average for their age and gender, in units of standard deviations.

```python
df['waz_score'].describe()
```

```python
x = df['waz_score']

color = (0.6, # redness
         0.4, # greenness
         0.8, # blueness
         0.6  # transparency
         )

edge_color = (0.2, # redness
              0.1, # greenness
              0.4, # blueness
              1  # transparency
              )

plt.hist(x, 20, edgecolor=edge_color, color=color)
plt.show()

plt.boxplot(x)
plt.show()
```

## Analysing Child's Age in Month

```python
df['child_age_months'].describe()
```

```python
x = df['child_age_months']

plt.hist(x, 6, edgecolor=edge_color, color=color)
plt.show()
```

## Analysing Birth Weight

```python
x = df['birth_weight_known'].value_counts()

plt.barh(x.index, x.values, edgecolor=edge_color, color=color)
plt.yticks(x.index, ['unknown', 'recorded'])
plt.show()
```

## Analysing Wealth Index

```python
x = df['wealth_index']
y = x.value_counts().sort_index()

plt.bar(y.index, y.values, color=color, edgecolor=edge_color)
plt.show()
```

## Analysing urban_rural

```python
df['urban_rural'].value_counts(normalize=True)
```

```python
x = df['urban_rural'].value_counts()

dark_purple = (0.4, # redness
                0.2, # greenness
                0.6, # blueness
                0.8  # transparency
                )

plt.pie(x, labels=['urban', 'rural'], startangle = 90, colors=[color, dark_purple], 
        wedgeprops={'edgecolor': edge_color, 'linewidth': 2})
plt.show()
```

## Analysing child's gender

```python
x = df['child_sex'].value_counts()

bars = plt.bar(x.index, x.values, color=color, edgecolor=edge_color)
plt.bar_label(bars)
plt.xticks([0, 1], ['Female', 'Male'])
plt.show()
```

## Analysing Mother's Education

```python
x = df['mother_education'].value_counts()
x = x[x.index != 9.0]

plt.barh(x.index, x.values, color=color, edgecolor=edge_color)
plt.show()

df['mother_education'].value_counts()
```

## Analysing Province

```python
x = df['province'].value_counts()

plt.barh(x.index, x.values, color=color, edgecolor=edge_color)
plt.bar_label(bars)
plt.yticks([0, 1, 2], ['Balochistan', 'KPK', 'Sindh'])
plt.show()
```

## Analysing Disability

```python
df['disability'].value_counts(normalize=True)
```

```python
x = df['disability'].value_counts()

# keep it 0 or 1, claude says
plt.barh(x.index, x.values, color=color, edgecolor=edge_color)
plt.yticks(x.index, ['not asked', 'disability', 'no disability'])
plt.show()
```

## Analysing Stunted Distribution

```python
fig, ax = plt.subplots(figsize=(6, 5))
counts = df['stunted'].value_counts()
labels = ['Normal', 'Stunted']
ax.pie(counts, labels=labels, colors=[color, dark_purple], autopct='%1.1f%%',
       startangle=90, wedgeprops={'edgecolor': edge_color, 'linewidth': 2})
ax.set_title('Overall Stunting Distribution', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig("screenshots/ss_univariate/stunting_univariate_pie.png")
plt.show()
```
