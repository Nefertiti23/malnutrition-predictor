```python
import pandas as pd
df = pd.read_csv("../../data/raw/hh_combined.csv", low_memory=False)
print(f"Rows: {df.shape[0]:,} | Columns: {df.shape[1]}")
```

```python
df.head()
```

```python
cols_to_keep = [
    'HH1', 'HH2',          # Merge keys (Cluster & Household)
    'HC3',                 # Main material of the roof/walls (Housing Quality)
    'WS1',                 # Source of drinking water
    'WS8',                 # Type of toilet facility
    'helevel',             # Education level (Household head/generalized)
    'HH6',                 # Urban/Rural
    'HH7',                 # District
    'windex5',             # Wealth index
    'province'             # Province
]

hh_clean = df[cols_to_keep].copy()
hh_clean.columns = [
    'cluster_id', 'household_id', 
    'housing_quality', 'water_source', 
    'sanitation_type', 'education_level', 
    'urban_rural', 'district', 
    'wealth_index', 'province'
]
```

```python
# Fill education and sanitation with the most common value (mode)
hh_clean['education_level'] = hh_clean['education_level'].fillna(hh_clean['education_level'].mode()[0])
hh_clean['sanitation_type'] = hh_clean['sanitation_type'].fillna(hh_clean['sanitation_type'].mode()[0])

# 2. Encode 1 for urban 2 for rural
hh_clean['urban_rural'] = hh_clean['urban_rural'].map({1.0: 0, 2.0: 1})

# 3. Encode Province
hh_clean['province'] = hh_clean['province'].map({
    'Balochistan': 0,
    'KPK': 1,
    'Sindh': 2
})

# 4. Final verification
print(f"Final shape: {hh_clean.shape}")
print(hh_clean.isnull().sum())
```

```python
# Fill housing_quality and water_source with the most common value
hh_clean['housing_quality'] = hh_clean['housing_quality'].fillna(hh_clean['housing_quality'].mode()[0])
hh_clean['water_source'] = hh_clean['water_source'].fillna(hh_clean['water_source'].mode()[0])

# Verify the fix
print(hh_clean.isnull().sum())
```

```python
hh_clean.to_csv("../../data/processed/hh_cleaned.csv", index=False)
print("Saved!")
```
